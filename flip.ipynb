## Cleanup

```python
"""
1 - Drop duplicate values using pid
2 - Both product rating and overall rating column are same we delete one 
3 - we dont need product url and image so we dropp'em
4 - expanding product category tree and putting it in df
5 - changing data type of crawl_date to date
6 - Product rating name change and data type change



"""
```

#### Dropping duplicates

```python
df.drop_duplicates(subset='pid', inplace = True)
df.reset_index(drop=True, inplace=True)
df.info()
```

#### Dropping image and overall rating

```python
df.drop(columns = ['image', 'overall_rating', 'product_url','uniq_id'], inplace = True)
```

```python
df.info()
```

#### Individual Category

```python
df['product_category_tree'].unique()
df['product_category_tree'] =  df['product_category_tree'].str.replace('["', '').str.replace('"]','')
categories = df['product_category_tree'].str.split(' >> ', expand=True)
categories

df['main_category'] = categories[0]
df['sub_category'] = categories[1]

df[['product_name','main_category','sub_category']].head(10)
```

#### product rating name change to rating and datatype change

```python

df.rename(columns={'product_rating':'Rating'}, inplace=True)
df['Rating'] = df['Rating'].replace('No rating available', np.nan)
df['Rating'] = df['Rating'].astype(float)
df['Rating'].unique()
```

#### Date Fix

```python
df.rename(columns={'crawl_timestamp': 'date'}, inplace=True)
df['date'] = pd.to_datetime(df['date'])
df.info()
```

#### Both price columns checking


```python
df['retail_price'].isnull().sum()
df['discounted_price'].isnull().sum()

# Both are 78 so ig both are null at same index

# Dropping those rows where value is null

df.dropna(subset='retail_price', inplace=True)
df['retail_price'].isnull().sum()
df['discounted_price'].isnull().sum()


# so its doneeeeeeeeeeeeeeee
```

## KPIs

```python
"""
IMP KPIs that we can find

1 - Top ten categories
2 - Which Brand generates most avg money 
3 - We can check monthly sales here and which category and brand is at top every month in avg sales 
4 - We can split the category column to check which category hast most sales.
5 - is_fk_advantage = flipkart advantage or now flipkart assured where the product goes through strict checking and all
    i) We can use it to check wheather fk products have more rating or not
    ii) Do these products have more price or more discount
    iii) Which category has most fk_advantage

"""
```

```python
# Top ten categories 

df.groupby('main_category')['discounted_price'].sum().sort_values(ascending=False).head(10).reset_index()


# Couting total products in each category
# df.groupby('main_category')['pid'].count().reset_index().sort_values(by = 'pid',ascending=False).head(10)

```

```python
# Which Brand generates most avg money and thier quantity

df.groupby(['brand','main_category']).agg(
    revenue = ('discounted_price','mean'),
    quantity =  ('pid','count')
).sort_values(by = 'revenue' ,ascending=False).head(10).reset_index()

# Count
# df.groupby('brand')['pid'].count().reset_index().sort_values(by = 'pid',ascending=False).head(10)
```

```python
# With those which have higher quantity (quantity > 10)
df.groupby(['brand','main_category']).agg(
    revenue = ('discounted_price','mean'),
    quantity =  ('pid','count')
).query('quantity >= 10').sort_values(by = 'revenue' ,ascending=False).head(10).reset_index()
```

```python
# Monthly sales and which category and brand is at top every month in avg sale
# lets add a year month column in our df

df['year_month'] = df['date'].dt.strftime('%Y-%m')

df.groupby('year_month')['discounted_price'].sum().reset_index().sort_values(by = 'year_month').head(10)
```

```python
"""
To get the monthly sale and the brand on top we have to create two tables one is monthly
sales and top category sales then we merge these two
"""

monthly_sales = df.groupby('year_month')['discounted_price'].sum().reset_index().sort_values(by = 'year_month').head(10).rename(columns={'discounted_price':'Total Revenue'})

# Top category per month
top_cat = df.groupby(['year_month','main_category'])['discounted_price'].sum().reset_index().sort_values(['year_month','discounted_price'], ascending=[True, False]).groupby('year_month').first().reset_index().rename(columns={"main_category":"Top Category", "discounted_price": "Category Revenue"})

Top_cat_month = pd.merge(monthly_sales, top_cat, on='year_month')
Top_cat_month
```

```python
# Flipkart Advantage or assured
"""
    i) We can use it to check wheather fk products have more rating or not
    ii) Do these products have more price or more discount
    iii) Which category has most fk_advantage
"""

df['is_FK_Advantage_product'].value_counts()
```

```python
# i) We can use it to check wheather fk products have more rating or not

df.groupby('is_FK_Advantage_product')['Rating'].mean().reset_index()
```

```python
# ii) Do these products have more price or more discount

# lets create a column discount percentage
df['dis_per'] = (((df['retail_price'] - df['discounted_price'])/df['retail_price'])*100).round(2)
df['dis_per']

df.groupby('is_FK_Advantage_product').agg(
    avg_price = ('retail_price','mean'),
    avg_discount_percent =  ('dis_per','mean')
).reset_index()
```

```python
# iii) Which category has most fk_advantage

df.groupby('main_category')['is_FK_Advantage_product'].sum().sort_values(ascending=False).head(10).reset_index()
```
